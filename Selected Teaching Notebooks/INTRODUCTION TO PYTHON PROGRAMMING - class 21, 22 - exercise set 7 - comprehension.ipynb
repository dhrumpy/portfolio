### Q 9:

Given the list:

    numbers = [4, -2, 7, -9, 0, 5]

replace every negative number with 0 and keep the rest.

```python
# do not run this cell:
```

```python
# write your code here: 
numbers = [4, -2, 7, -9, 0, 5]

res = []
for n in numbers:
    if n < 0:
        res.append(0)
    else:
        res.append(n)
res
```

```python
[n for n in numbers if n >= 0]
```

```python
[n if n >= 0 else 0 for n in numbers]
```

```python
numbers = [4, -2, 7, -9, 0, 5]

res = []

for n in numbers:
    if n < 0:
        res.append("-")
    elif n > 0:
        res.append("+")
    else:
        res.append("ZERO")
res
```

```python
["+" if n >0 else "-" if n < 0 else "ZERO" for n in numbers]
```
