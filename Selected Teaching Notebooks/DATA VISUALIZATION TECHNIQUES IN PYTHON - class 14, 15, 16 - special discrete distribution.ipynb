```python
x, l = symbols("x, lambda", real=True, positive=True)

f = exp(-l*x)
N = integrate(f, (x, 0, oo))

pdf_fun = f/N

integrate(pdf_fun, (x, 0, oo))
```

```python
x, l = symbols("x, lambda", real=True, positive=True)

f = exp(-l*x)
N = integrate(f, (x, 0, oo))

pdf_fun = f/N


l_val = solve(integrate(x*pdf_fun, (x, 0, oo)) - 4)[0]
l_val
```

```python
x = symbols("x", real=True, positive=True)

f = exp(-0.2*x)
N = integrate(f, (x, 0, oo))

pdf_fun = f/N

pdf_fun
```

```python
ns = [20, 200, 2000]

k = list(range(0, 20))
l = 5

fig, ax = plt.subplots(nrows=3, ncols=1)
ax = ax.flatten()
for i, n in enumerate(ns):
    p = l/n
    sns.barplot(binom.pmf(k=k, n=n, p=p), label="binomial", ax=ax[i])
    sns.barplot(poisson.pmf(k=k, mu=n*p), label="poisson", ax=ax[i])
    ax[i].set_title(f"n={n}")
    
    ax[i].legend()

plt.tight_layout()
```
