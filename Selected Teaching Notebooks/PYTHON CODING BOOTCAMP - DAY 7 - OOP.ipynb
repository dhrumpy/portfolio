---------------------------------------------------------------------------------
- Add a method `compare_memory(self, other)` that compares the memory of two mobiles and prints which one has more memory.

- Add a method `is_same_price(self, other)` that returns `True` if prices are equal, else `False`

- Add a method `apply_discount(self, percent)` to reduce the price by given percent





````python
# do not run this cell
class Mobile:
    pass


mobile1 = Mobile("Samsung", "Galaxy A52", 128, 40000)
mobile2 = Mobile("Apple", "iPhone 12", 64, 80000)
print(mobile1.compare_memory(mobile2))
print(mobile1.is_same_price(mobile2))
print(mobile1.apply_discount(10))
````

````python
# write your code here:
# do not run this cell
class Mobile:
    def __init__(self, brand, model, memory, price):
        self.brand = brand
        self.model = model
        self.memory = memory
        self.price = price

    def compare_memory(self, other):
        if self.memory > other.memory:
            return f"{self.brand} {self.model} has more memory than {other.brand} {other.model}"

    def is_same_price(self, other):
        return self.price == other.price

    def apply_discount(self, discount):
        return round(self.price*(1 - (discount/100)), 2)
            
    


mobile1 = Mobile("Samsung", "Galaxy A52", 128, 40000)
mobile2 = Mobile("Apple", "iPhone 12", 64, 80000)
print(mobile1.compare_memory(mobile2))
print(mobile1.is_same_price(mobile2))
print(mobile1.apply_discount(10))
````

### Q 2: Student Class and Academic Evaluation

- Define a `Student` class with attributes: `name`, `student_id`, `grades` (a dictionary of subjects and marks), and `year`.

- Create two student objects and print their details:
  - `s1 = Student("Aiko Tanaka", "S123", {"math": 88, "english": 92, "science": 81}, 2)`
  - `s2 = Student("Hiro Yamamoto", "S456", {"math": 75, "english": 85, "science": 89}, 2)`

---------------------------------------------------------------------------------
- Add a method `average_grade(self)` that returns the average grade across all subjects.

- Add a method `top_subject(self)` that returns the subject with the highest mark.

- Add a method `compare_student(self, other)` that prints the name of the student with the higher average grade.


````python
# do not run this cell:
class Student():
    pass
        

s1 = Student("Aiko Tanaka", "S123", {"math": 85, "english": 77, "science": 85}, 2)
s2 = Student("Hiro Yamamoto", "S456", {"math": 75, "english": 90, "science": 89}, 2)
print(s1.average_grade())
print(s2.average_grade())
print(s1.top_subject())
print(s2.top_subject())
print(s1.compare_student(s2))
````

````python
# write your code here:
class Student():
    def __init__(self, name, student_id, grades, year):
        self.name = name
        self.student_id = student_id
        self.grades = grades
        self.year = year

    def average_grade(self):
        return round(sum(list(self.grades.values()))/len(self.grades.values()), 2)

    def top_subject(self):
        return ", ".join([k.capitalize() for k, v in self.grades.items() if v == max(list(self.grades.values()))])
         
        

s1 = Student("Aiko Tanaka", "S123", {"math": 85, "english": 77, "science": 85}, 2)
s2 = Student("Hiro Yamamoto", "S456", {"math": 75, "english": 90, "science": 89}, 2)
print(s1.average_grade())
print(s2.average_grade())
print(s1.top_subject())
print(s2.top_subject())
# print(s1.compare_student(s2))
````
