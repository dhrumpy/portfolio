```python
image = plt.imread('starry night.jpg')
plt.imshow(image)
image.shape
```

```python
A = image[:, :, 0]
plt.imshow(A)
U, S, VT = np.linalg.svd(A)
len(S)
```

```python
U.shape, VT.shape
```

```python
k = 50
image_compressed = U[:, :k] @ np.diag(S[:k]) @ VT[:k, :]
plt.imshow(image_compressed)
```

```python
U[:, :k].shape
```

```python
k = 400
image_compressed = np.zeros_like(image)
for i in range(3):
    A = image[:, :, i]
    U, S, VT = np.linalg.svd(A)
    image_compressed[:, :, i] = U[:, :k] @ np.diag(S[:k]) @ VT[:k, :]
    
plt.imshow(image_compressed)
```
