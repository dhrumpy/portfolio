## Q 1:

Write a Python function that calculates and prints the Fibonacci series

```python
# do not run this cell:
def fibonacci(n):
    if n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    else:
        fib_series = [0, 1]
        for i in range(2, n):
            fib_series.append(fib_series[i-1] + fib_series[i-2])
        return fib_series
    

print(fibonacci(5))
print(fibonacci(10))
print(fibonacci(16))
```

## Q 2:

Write a Python function, `diagonal_difference(arr)`, that accepts a square matrix (or a 2-d array/list ) and returns the absolute difference between the sum of diagonals

```python
# do not run this cell:
def diagonal_difference(arr):
    for size in [len(x) for x in arr]:
        if len(arr) != size:
            return 'This is not a square matrix!'
    diagonal1, diagonal2 = 0, 0
    for i in range(len(arr)):
        diagonal1 += arr[i][i]
        diagonal2 += arr[i][len(arr)-1-i]
    return diagonal2 - diagonal1

print(diagonal_difference([[1, 2, 3],[4, 5, 6], [7, 8, 9]]))      
print(diagonal_difference([[1, 2, 3],[4, 5, 6], [9, 8, 9]])) 
print(diagonal_difference([[11, 2, 4], [4, 5, 6], [10, 8, -12]]))
print(diagonal_difference([[1, 2], [3, 4], [5, 6]])) 
```

## Q 3:

`Users on longer flights like to start a second movie right when their first one ends, but they complain that the plane usually lands before they can see the ending. So you're building a feature for choosing two movies whose total runtimes will equal the exact flight length.`

Write a Python function that takes an integer flight_length (in minutes) and a list of integers movie_lengths (in minutes) and returns a boolean indicating whether there are two numbers in movie_lengths whose sum equals flight_length.

```python
# do not run this cell:
def can_watch_two_movies(flight_length, movie_lengths):
    for i, t1 in enumerate(movie_lengths):
        for j, t2 in enumerate(movie_lengths):
            if (i != j) and (t1 + t2) == flight_length:
                return True
    return False
        

print(can_watch_two_movies(120, [80, 90, 60, 70, 50, 30]))
print(can_watch_two_movies(150, [60, 70, 80, 90, 100]))
print(can_watch_two_movies(160, [40, 50, 60, 70, 80]))
```

## Q 4: 

Write a Python function that selects all words that have all the same vowels (in any order and/or number) as the first word, including the first word

```python
# do not run this cell:
def get_vowels(word):
    vowels = 'a e i o u'.split()
    res = []
    for char in word:
        if char in vowels:
            res.append(char)
    return set(res)

def same_vowels(arr):    
    first_word = arr[0]
    res = [first_word]
    for word in arr[1:]:
        if get_vowels(first_word) == get_vowels(word):
            res.append(word)
    return res
            
            
# print(same_vowels(["toe", "ocelot", "maniac"]))
print(same_vowels(["many", "carriage", "emit", "apricot", "animal"]))
# print(same_vowels(["hoops", "chuff", "bot", "bottom"]))
```

## Q 5:

Write a Python function, `reverse_integer(integer)`, that accepts a signed integer and returns an integer with its digits reversed. Also, check if both the input and the output lie in the range $[-10^6, 10^6]$; otherwise, return 0. Your code should also return 0 if the input is not an integer object

`Note`: Try this problem in following two approaches:
 
         1) Using string manipulation 
         2) Using '%' and '//' operators  

```python
#  do not run this cell:
# using string manipulation:

def get_digits(n):
    res = ''
    for digit in str(n)[::-1]:
        res += digit
    return int(res)
    
def reverse_integer(n):
    if type(n) != int:
        return 0
    if (n < -1e6) or (n > 1e6):
        return 0
    else:
        if n < 0:
            return (-1)*get_digits(str(n)[1:])
        else:
            return get_digits(str(n))
            
            

print(reverse_integer(-839))
print(reverse_integer(1234))
print(reverse_integer(232456357))
print(reverse_integer('hello'))
```

```python
#  do not run this cell:
# using '%' and '//' operators:

def get_reversed_digits(n):
    reversed_num = 0
    while n != 0:
        reversed_num = reversed_num*10 + n%10
        n //= 10
    return reversed_num
    
def reverse_integer(n):
    if type(n) != int:
        return 0
    if (n < -1e6) or (n > 1e6):
        return 0
    else:
        if n < 0:
            return (-1)*get_reversed_digits((-1)*n)
        else:
            return get_reversed_digits(n)
            
            

print(reverse_integer(-839))
print(reverse_integer(1234))
print(reverse_integer(232456357))
print(reverse_integer('hello'))
```

## Q 6:

Write a Python function to check if two strings are anagrams of each other (contain the same characters in different orders)

`Note`: Try this problem in following two approaches:
 
         1) Using nested loops
         2) Using dictionary 

```python
# do not run this cell:
# using nested for-loops:
def is_anagrams(s1, s2):
    if len(s1) != len(s1):
        return False
    else:
        for x in s1:
            flag = False
            for y in s2:
                if x == y:
                    flag = True
                    break
            if not flag:
                return False
        return True
    
print(is_anagrams('listen', 'silent'))
print(is_anagrams('angel', 'glean')) 
print(is_anagrams('clean', 'glean')) 
```

```python
# do not run this cell:
# using dictionary:
def get_freq(s1):
    freq = {k: 0 for k in s1}
    for s in s1:
        if s in freq.keys():
            freq[s] += 1
        else:
            freq[s] = 1
    return freq
        
def is_anagrams(s1, s2):
    if len(s1) != len(s1):
        return False
    else:
        return get_freq(s1) == get_freq(s2)
        
    
print(is_anagrams('listen', 'silent'))
print(is_anagrams('angel', 'glean')) 
print(is_anagrams('clean', 'glean')) 
```
