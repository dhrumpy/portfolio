```python
from scipy.stats import norm, t
import matplotlib.pyplot as plt

import numpy as np
from sympy import *

import seaborn as sns
sns.set_theme(style="whitegrid")
```

```python
z = symbols("z")
f = exp(-0.5*z**2)
N = integrate(f, (z, -oo, oo))

pdf = f/N

pdf
```

```python
z_cric = float(solve(integrate(pdf, (z, -z, z)) - 0.95)[0])
z_cric
```

```python
z = np.linspace(-5, 5, 100)
plt.plot(z, norm.pdf(z), label="normal")

dofs = [2, 10, 100]
for dof in dofs:
    plt.plot(z, t.pdf(z, df=dof), label=f"t: dof={dof}")

plt.legend()
```

```python
n, sample_mean, sample_std = 8, 12.05, 0.18

t_cric = t.ppf(0.975, df=n-1)

se = sample_std/np.sqrt(n)

ci_lower, ci_upper = sample_mean-se*t_cric, sample_mean+se*t_cric

(ci_lower, ci_upper)
```
