```python
# E1: die1 shows 6
# E2: die2 shows {3, 4, 5, 6}

# P(E1 and E2) = P(E1) x P(E2|E1)

(4/36)/(1/6)
```

```python
N = 100000
n1 = 0 # counts of die1 showing 6
n2 = 0 # counts of die1 showing 6 and die2 showing {3, 4, 5, 6}

for _ in range(N):
    if np.random.choice(a=[True, False], p=[1/6, 1-(1/6)]):
        n1 += 1
    if np.random.choice(a=[True, False], p=[4/36, 1-(4/36)]):
        n2 += 1

n2/n1
```

```python
count_has_disease_and_tested_positive = 0
count_tested_positive = 0

N = 100000

for _ in range(N):
    has_disease = np.random.choice(a=[True, False], p=[0.5/100, 1-(0.5/100)])
    if has_disease:
        tested_positive = np.random.choice(a=[True, False], p=[98/100, 1-(98/100)])
        if tested_positive:
            count_has_disease_and_tested_positive += 1
            count_tested_positive += 1
    else:
        tested_positive = np.random.choice(a=[True, False], p=[3/100, 1-(3/100)])
        if tested_positive:
            count_tested_positive += 1

count_has_disease_and_tested_positive/count_tested_positive
```
